```python
#0086 - Tính chu vi và diện tích của tam giác có sử dụng hàm
import math
def chuvi(a, b, c):
    return float(a + b + c)
def dientich(a, b, c):
    p = chuvi(a, b, c) / 2
    return math.sqrt(p * (p - a) * (p - b) * (p - c))
def main():
    a, b, c = map(int, input().split())
    cv = chuvi(a, b, c)
    dt = dientich(a,b,c)
    print(f"{cv:.1f} {dt:.3f}")
main()
```

```python
#0089 - Viết hàm tính n!
import math
def giaithua(n):
    return math.factorial(n)
def tinhc(n,k):
    return giaithua(n) // (giaithua(k) * giaithua(n-k))
def main():
    m = int(input())
    tong = 0
    for i in range(m):
        n, k = map(int, input().split())
        tong += tinhc(n,k)
    print(tong)
main()
```

```python
#0090 - Viết hàm P(x,n) để tính giá trị x^n
def P(x,n):
    return x**n
def main():
    k, x = map(int, input().split())
    F = 0
    for i in range(k):
        a, n = map(int, input().split())
        F += a*P(x,n)
    print(F)
main()
```

```python
#0091 - Viết chương trình vẽ hình chữ nhật bằng dấu * sử dụng chương trình con
def vehcn(d,r):
    if r <= 0 or d <= 0:
        return
    print("*" * d)
    for i in range(r - 2):
        if d > 1:
            print("*" + " " * (d-2) + "*")
        else:
            print("*")
    if r > 1:
        print("*" * d)
def main():
    d, r = map(int, input().split())
    vehcn(d,r)
main()
```

```python
#0092 - Sử dụng hàm lồng nhau
def cong(a,b):
    return a + b
def chia(a,b):
    return a / b
def main():
    a, b, c = map(int,input().split())
    congabc = int(cong(cong(a,b), c))
    congchia = cong(chia(a,b),c)
    chiacong = chia(a, cong(b,c))
    print(f"{congabc} {congchia:.2f} {chiacong:.3f}")
main()
```

```python
#Sử dụng hàm tìm số nhỏ nhất trong 2 số để tìm số nhỏ nhất trong 3 số
def min2so(a,b):
    return min(a,b)
def main():
    a, b, c = map(int, input().split())
    min3so = min2so(min2so(a,b), c)
    print(min3so)
main()
```

```python
#0094 - Viết hàm cộng trừ nhân chia 2 phân số
import math
def ucln(a, b):
    return math.gcd(a,b)
def phanso(tu,mau):
    g = ucln(abs(tu), abs(mau))
    tu //= g
    mau //= g
    if tu > mau and mau != 1:
        nguyen = tu // mau
        tumoi = tu % mau
        print(f"{nguyen},{tumoi}/{mau}")
    else:
        print(f"{tu}/{mau}")
def cong(a,b,c,d):
    phanso(a*d + c*b, b*d)
def tru(a,b,c,d):
    phanso(a*d - c*b, b*d)
def nhan(a,b,c,d):
    phanso(a*c, b*d)
def chia(a,b,c,d):
    phanso(a*d, b*c)
def main():
    a,b = map(int, input().split())
    c,d = map(int, input().split())
    cong(a,b,c,d)
    tru(a,b,c,d)
    nhan(a,b,c,d)
    chia(a,b,c,d)
main()
```

```python
#0095 - Viết chương trình con tính tổng các phần tử trong 1 mảng
def tinhtong(s):
    return sum(s)
def main():
    n = int(input())
    s = list(map(int, input().split()))
    tong = tinhtong(s)
    tb = tong / n
    print(f"{tb:.1f}")
main()
```

```python
#0096 - Viết chương trình con tính tổng các phần tử chẵn trong 1 mảng
def tinhtong(s):
    tong = 0
    count = 0
    for x in s:
        if x % 2 == 0:
            tong += x
            count += 1
    return tong, count
def main():
    n = int(input())
    s = list(map(int, input().split()))
    tongchan, demchan = tinhtong(s)
    if demchan > 0:
        tb = tongchan / demchan
        print(f"{tb:.2f}")
main()
```

```python
#0097 - Đếm xem trong mảng có bao nhiêu số nguyên tố có sử dụng hàm
def nguyento(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
def main():
    n = list(map(int, input().split()))
    count = 0
    for x in n:
        if nguyento(x):
            count += 1
    if count == 0:
        print("-")
    else:
        print(count)
main()
```

```python
#0098 - Tính trung bình cộng các số nguyên tố trong mảng có sử dụng hàm
def nguyento(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
def main():
    n = list(map(int, input().split()))
    tong = 0
    count = 0
    for x in n:
        if nguyento(x):
            tong += x
            count += 1
    if count == 0:
        print("-")
    else:
        tb = tong / count
        print(f"{tb:.2f}")
main()
```

```python
#0099 - Tính trung bình cộng các số chính phương trong mảng có sử dụng hàm
def chinhphuon(n):
    if n < 0:
        return False
    for i in range(n+1):
        if i * i == n:
            return True
        if i * i > n:
            break
    return False
def main():
    n = list(map(int, input().split()))
    tong = 0
    count = 0
    for x in n:
        if chinhphuon(x):
            tong += x
            count += 1
    if count == 0:
        print("-")
    else:
        tb = tong / count
        print(f"{tb:.1f}")
main()
```

```python
#0100 - Viết 2 chương trình con tìm Max, Min và vị trí của Max, Min trong mảng
def findmax(s):
    giatrimax = max(s)
    vitri = [i for i in range(len(s)) if s[i] == giatrimax]
    return giatrimax, vitri
def findmin(s):
    giatrimin = min(s)
    vitri = [i for i in range(len(s)) if s[i] == giatrimin]
    return giatrimin, vitri
def main():
    n = int(input())
    s = list(map(int, input().split()))
    giatrimax, vitrimax = findmax(s)
    giatrimin, vitrimin = findmin(s)
    print(giatrimax, *vitrimax)
    print(giatrimin, *vitrimin)
main()
```
