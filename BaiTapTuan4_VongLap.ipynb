```python
#W4A6
import sys
for line in sys.stdin:
    if not line.strip():
        continue
    a,b = map(int, line.split())
    tong = 0
    for i in range(a, b+1):
        nguyento = True
        if i < 2:
            nguyento = False
        else:
            for x in range(2, i):
                if i % x == 0:
                    nguyento = False
        if nguyento:
            tong += i
    print(tong)
```

```python
#W4A7
n = int(input())
maxnguyento = 0
for i in range(1,n+1):
    uoc = False
    nguyento = True
    if n % i == 0:
        uoc = True
    if uoc:
        if i < 2:
            nguyento = False
        else:
            for x in range(2,i):
                if i % x == 0:
                    nguyento = False
        if nguyento:
            if i > maxnguyento:
                maxnguyento = i
print(maxnguyento)
```

```python
#W4A8
n = int(input())
count = 0
while str(n) != str(n)[::-1]:
    reverse = int(str(n)[::-1])
    n = n + reverse
    count += 1
print(count, n)
```

```python
#W4A9
n = int(input())
lst = []
for x in range(0, int(n**0.5)+1):
    cp = x*x
    s = str(cp)
    if len(set(s)) == len(s):
        lst.append(s)
if len(lst) > 0:
    print(" ".join(lst))
else:
    print("(no number)")
```

```python
#W4A10
n = int(input())
x = 1
maxlen = 0
for i in range(1,n+1):
    k = i
    length = 1
    while k != 1:
        if k % 2 == 0:
            k //= 2
        else:
            k = 3*k + 1
        length += 1
    if length > maxlen:
        maxlen = length
        x = i
print(x, maxlen)
```

```python
#W4A13
def tonguoc(n):
    if n <= 1:
        return 0
    tong = 1
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            tong += i
            if i != n // i:
                tong += n // i
    return tong
a, b = map(int, input().split())
if tonguoc(a) == b and tonguoc(b) == a:
    print("true")
else:
    print("false")
```
