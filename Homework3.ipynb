```python
#Bài 4
h = int(input())
day = 2*h - 1
trong = 1
lui = h - 2
print(" " * (lui + 1) + "*")
if h > 1:
    for i in range(h-2):
        print(f"{lui * " "}*{trong * " "}*")
        lui -= 1
        trong += 2
    print("*" * day)
```

```python
#Bài 14
x = int(input())
if x == -1:
    print("Chưa nhập số nào")
else:
    maxval = x
    minval = x
    while True:
        x = int(input())
        if x == -1:
            break
        if x > maxval:
            maxval = x
        if x < minval:
            minval = x
print(maxval, minval)
```

```python
#Bài 20
n = input()
count = 0
for i in n:
    d = int(i)
    if d >= 2:
        for x in range(2, int(d**0.5)+1):
            if d%x == 0:
                break
        else:
            count += 1
print(count)
```
