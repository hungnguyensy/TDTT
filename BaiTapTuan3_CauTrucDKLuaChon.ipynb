```python
#W3A16
x = float(input())
tronxuong = int(x // 1)
if x % 1 == 0:
    tronlen = tronxuong
else:
    tronlen = tronxuong + 1
thapphan = x - tronxuong
if x >= 0:
    if thapphan >= 0.5:
        gannhat = tronxuong + 1
    else:
        gannhat = tronxuong
else:
    if thapphan > 0.5:
        gannhat = tronxuong + 1
    else:
        gannhat = tronxuong
print(tronlen, tronxuong, gannhat)
```
