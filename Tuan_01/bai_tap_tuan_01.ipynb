```python
#Bài 6
a = float(input("Nhập vào một số a: "))
b = float(input("Nhập vào một số b: "))
c = float(input("Nhập vào một số c: "))
if a > 0 and b > 0 and c > 0 and ((a + b) > c) and ((a + c) > b) and ((c + b) > a):
    p = ( a + b + c)/2
    dientich = (p * (p-a) * (p-b) * (p-c))**0.5
    print(f"diện tích tam giác có 3 cạnh là a,b,c là: {round(dientich, 1)}")
else:
    print("Khong phai 3 canh cua tam giac")
```

```python
#Bài 8
name = input("Nhập vào họ tên chủ hộ: ")
thangtruoc = int(input("Nhập chỉ số điện kế tháng trước: "))
thangnay = int(input("Nhập chỉ số điện kế tháng này: "))
kw = thangnay - thangtruoc
if kw >= 0 and kw <= 50:
    tien = kw * 1984
elif kw >= 51 and kw <= 100:
    tien = 50*1984 + (kw - 50)*2050
elif kw >= 101 and kw <=200:
    tien = 50*1984 + 50*2050 + (kw - 100)*2380
elif kw >= 201 and kw <= 300:
    tien = 50*1984 + 50*2050 + 100*2380 + (kw - 200)*2998
elif kw >= 301 and kw <= 400:
    tien = 50*1984 + 50*2050 + 100*2380 + 100*2998 + (kw - 300)*3350
else:
    tien = 50*1984 + 50*2050 + 100*2380 + 100*2998 + 100*3350 + (kw - 400)*3460
tienphaitra = round(tien * 1.08)
print(f"Ho va ten: {name}")
print(f"Tien phai tra la: {tienphaitra}")
```

```python
#Một số bài tập khác - Bài 7: Tính điểm học lực
diem = float(input("Nhập điểm từ 0 - 10: "))
while diem < 0 or diem > 10:
    diem = float(input("Điểm phải từ 0 - 10, hãy nhập điểm khác: "))
if diem < 5:
    print("học lực yếu")
elif diem >= 5 and diem < 6.5:
    print("học lực trung bình")
elif diem >= 6.5 and diem < 8:
    print("học lực khá")
else:
    print("học lực giỏi")
```
